# 🟠 Medium: 2D Convolution

Implement **2D convolution** from scratch.

### Signature
```python
def my_conv2d(x, weight, bias=None, stride=1, padding=0):
    # x: (B, C_in, H, W), weight: (C_out, C_in, kH, kW)
    # Returns: (B, C_out, H_out, W_out)
```

### Rules
- Do NOT use `F.conv2d` or `nn.Conv2d`
- Support `stride` and `padding` parameters
- `F.pad` for zero-padding is allowed

````python
import torch
import torch.nn.functional as F
import math
````

````python
# ✏️ YOUR IMPLEMENTATION HERE

def my_conv2d(x, weight, bias=None, stride=1, padding=0):
    pass  # extract patches, apply kernel, handle stride/padding

    batch_size = x.shape[0]
    origin_height = x.shape[2]
    origin_width = x.shape[3]

    c_out = weight.shape[0]
    kh = weight.shape[2]
    kw = weight.shape[3]

    result_height = math.ceil((origin_height + 2 * padding - kh + 1) / stride)
    result_width = math.ceil((origin_width + 2 * padding - kw + 1) / stride)


    if padding > 0:
      x = F.pad(x, (padding, padding, padding, padding), 'constant', 0)

    conv_result = torch.empty(batch_size, c_out, result_height, result_width)

    for l in range(batch_size):           #For each elem in the batch
      for f in range(c_out):              #For each output filter
        target_filter = weight[f]
        for i in range(0, result_height):
          for j in range(0, result_width):

            target_area = x[l, :, i*stride : i*stride+kh, j*stride : j*stride+kw]

            if bias is not None:
              conv_result[l][f][i][j] = torch.sum(target_area * target_filter) + bias[f]
            else:
              conv_result[l][f][i][j] = torch.sum(target_area * target_filter)


    return conv_result
````

````python
# 🧪 Debug
x = torch.randn(1, 3, 8, 8)
w = torch.randn(16, 3, 3, 3)
print('Output:', my_conv2d(x, w).shape)
print('Match:', torch.allclose(my_conv2d(x, w), F.conv2d(x, w), atol=1e-4))
````
